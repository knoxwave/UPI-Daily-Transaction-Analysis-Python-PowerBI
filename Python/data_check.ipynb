```python
import pandas as pd
import numpy as na

df = pd.read_csv("UPI_Daily_Statistics_Clean.csv")
df.head()
```

```python
# First and last day's transaction volume
start_value = df["Volume_Million"].iloc[0]
end_value = df["Volume_Million"].iloc[-1]

growth_percent = ((end_value - start_value) / start_value) * 100

print(f"Growth %: {growth_percent:.2f}%")
```
